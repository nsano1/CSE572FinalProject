```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("education_1980_2019.csv")

# Group by year and calculate mean percentage for each education level
yearly_education = df.groupby("Year")[
    ["Percent of adults with less than a high school diploma",
     "Percent of adults with a high school diploma only"]
].mean().reset_index()


plt.figure(figsize=(12, 6))
plt.plot(yearly_education["Year"],
         yearly_education["Percent of adults with less than a high school diploma"],
         label="Less than High School", marker='o')
plt.plot(yearly_education["Year"],
         yearly_education["Percent of adults with a high school diploma only"],
         label="High School Only", marker='s')

plt.title("Education Levels by Year (1980–2019)")
plt.xlabel("Year")
plt.ylabel("Percentage of Adults")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
